```python
import heapq

goal=[[1,2,3],
      [4,5,6],
      [7,8,0]]

def heuristic(graph):
    h=0
    for i in range(len(graph)):
        for j in range(len(graph[0])):
            if graph[i][j]!=0:
                x=(graph[i][j]-1)//3
                y=(graph[i][j]-1)%3

                dis=abs(i-x)+abs(j-y)
                h+=dis
    return h


def astar(graph):

    pq=[(heuristic(graph),0,graph,[graph])]
    visited=set()
    while pq:

        f,g,curr,path=heapq.heappop(pq)
        if curr==goal:
            return path
        key=str(curr)
        if key in visited:
            continue
        visited.add(key)

        # find the pos of 0

        for i in range(3):
            for j in range(3):
                if curr[i][j]==0:
                    x=i
                    y=j
        dir=[(-1,0),(0,-1),(1,0),(0,1)]

        for r,c in dir:
            nr=x+r
            nc=y+c

            # boundries
            if 0<=nr<3 and 0<=nc<3:
                new_matrix=[row[:] for row in curr]
                new_matrix[nr][nc],new_matrix[x][y]=new_matrix[x][y],new_matrix[nr][nc]

                if str(new_matrix) not in visited:
                    new_g=g+1
                    new_f=new_g+heuristic(new_matrix)

                    heapq.heappush(pq,(new_f,new_g,new_matrix,path+[new_matrix]))
    return None

path = astar(graph)
if path:
    for matrix in path:
        for row in matrix:
            print(row)
        print("\n")
    print(f"Found in {len(matrix)} steps")
else:
    print("Not found")
```

```python
# using adminssible heuristic distance
graph={}
n=int(input("ENter number of vertex:"))
vertex=[]
for i in range(n):
    val=input(f"Enter for vertex {i} :")
    vertex.append(val)
    graph[val]=[]

e=int(input("Enter no of edges : "))
for i in range(e):
    w,u,v=input("enter w u v").split()
    w=int(w)

    graph[u].append((v,w))
    graph[v].append((u,w))

heuristic={}
for i in vertex:
    dis=input(f"Enter heuristic dist of {i}: ")
    heuristic[i]=int(dis)
```

```python
import heapq

def astar(graph,heuristic,start,goal):
    pq=[(heuristic[start],0,start,[start])]

    visited=set()

    while pq:

        f,g,curr,path=heapq.heappop(pq)
        if curr==goal:
            return path,g
        if curr in visited:
            continue
        visited.add(curr)

        # check in neighour
        for neigh,weight in graph[curr]:
            if neigh not in visited:
                new_g=g+weight
                new_f=heuristic[neigh]+new_g

                heapq.heappush(pq,(new_f,new_g,neigh,path+[neigh]))
    return None,0

start=input("Enter start vertex :")
goal=input("Enter goal vertex :")

path,dis=astar(graph,heuristic,start,goal)
if path:
    print(path)
    print(dis)
else:
    print("Not found")
```
