```python
import heapq

def dijkstra(graph,start,n):

    distance=[999999]*n
    parent=[-1]*n

    pq=[(0,start)]
    distance[start]=0

    while pq:
        dis,u=heapq.heappop(pq)

        if distance[u]<dis:
            continue

        for v in range(n):
            if graph[u][v] != 0:
                new_dist=dis+graph[u][v]
    
                if new_dist<distance[v]:
                    distance[v]=new_dist
                    heapq.heappush(pq,(new_dist,v))
                    parent[v]=u
    # print distance here
    for i in range(n):
        if distance[i]==999999:
            print(f"Unreachable from {start} -> {i} ")
        else:
            path=[]
            curr=i
            while(curr!=-1):
                path.append((curr))
                curr=parent[curr]

            path.reverse()
            print("Path : ",path)
            print(f"Distance : {start} to {i}  :",distance[i])
```

```python
## using graph
import heapq

def dijkstra_graph(graph,start,n):

    distance={node:float('inf') for node in graph}
    parent={node:None for node in graph}

    pq=[(0,start)]
    distance[start]=0

    while pq:
        dis,u=heapq.heappop(pq)

        if distance[u]<dis:
            continue

        for neighbour,weight in graph[u]:
            new_dist=dis+weight
    
            if new_dist<distance[neighbour]:
                distance[neighbour]=new_dist
                heapq.heappush(pq,(new_dist,neighbour))
                parent[neighbour]=u
    # print distance here
    for vertex in graph:
        if distance[vertex]==float('inf'):
            print(f"Unreachable from {start} -> {vertex} ")
        else:
            path=[]
            curr=vertex
            while(curr!=None):
                path.append((curr))
                curr=parent[curr]

            path.reverse()
            print("Path : ",path)
            print(f"Distance : {start} to {vertex}  :",distance[vertex])

graph={}

v=int(input("Enter the number of vertex:"))

for i in range(v):
    print("Vertex: ",i)
    val=input("Enter value of vertex:")
    graph[val]=[]

e=int(input("Enter the number of egdes: "))

for i in range(e):
    w,u,v=input("Enter weight,u,v: ").split()

    w=int(w)

    graph[u].append((v,w))
    graph[v].append((u,w))

print(graph)

start=input("Enter the start vertex : ")

dijkstra_graph(graph,start,v)
```
