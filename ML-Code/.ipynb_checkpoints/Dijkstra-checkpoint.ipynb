```python
#Implement Dijkstra's algorithm to find the shortest path from a source vertex in a weighted graph. Display the shortest distances and
#paths.
import heapq

def dijkstra(graph,start,n):

    distance=[999999]*n
    parent=[-1]*n

    pq=[(0,start)]
    distance[start]=0

    while pq:
        dis,u=heapq.heappop(pq)

        if distance[u]<dis:
            continue

        for v in range(n):
            if graph[u][v] != 0:
                new_dist=dis+graph[u][v]
    
                if new_dist<dist[v]:
                    dist[v]=new_dist
                    heapq.heappush(pq,(new_dist,v))
                    parent[v]=u

    # print distance here
    for i in range(n):
        if dist[i]==999999:
            print(f"Unreachable from {start} -> {i} ")
        else:
            path=[]
            curr=i
            while(curr!=-1):
                path.append((curr))
                curr=parent[curr]

            print("Path : ",path)
            print(f"Distance : {start} to {i}  :",distance[i])


graph=[]
v=int(input("Enter the number of vertex : "))
print("Enter the graph:")

for i in range(v):
    row=list(map(int,input().split()))
    graph.append(row)

start=int(input("Enter start Vertex: "))

dijkstra(graph,start,v)    
```
