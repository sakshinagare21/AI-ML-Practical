```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans,AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram,linkage
```

```python
df=pd.read_csv("50_Startups.csv")
df.head(5)
```

```python
df['State'].value_counts()
```

```python
df['State']=df['State'].map({
    "New York":0,
    "California":1,
    "Florida":2
})

print(df.head(5))
```

```python
df=df.dropna()
```

```python
X=df
```

```python
scaler=StandardScaler()

x_scaled=scaler.fit_transform(X)
```

```python
# kmeans
# clustering
from sklearn.metrics import silhouette_score
wcss=[]

for i in range(2,11):
    kmeans=KMeans(n_clusters=i,n_init=10,random_state=42)
    kmeans.fit(x_scaled)
    labels=kmeans.fit_predict(x_scaled)

    score = silhouette_score(x_scaled, labels)
    print(score)

    wcss.append(kmeans.inertia_)

print(wcss)

#plot graph
plt.plot(range(2,11),wcss,marker='o')
plt.title("Elbow method")
plt.xlabel("Number of Cluster")
plt.ylabel("wcss")
plt.show()
```

```python
Kmeans=KMeans(n_clusters=3,n_init="auto",random_state=42)
kmeans.fit(x_scaled)
kmeans_labels=kmeans.fit_predict(x_scaled)

print(silhouette_score(x_scaled, kmeans_labels))
```

```python
# plt.scatter(X['R&D Spend'],X['Profit'],c=kmeans_labels,cmap="rainbow")
# plt.show()

centroids=kmeans.cluster_centers_

plt.scatter(x_scaled[:,0],x_scaled[:,1],c=kmeans_labels,cmap="rainbow")
plt.scatter(centroids[:, 0],
    centroids[:, 1],
    color='black',
    marker='X',
    s=250,
    label='Centroids'
)
plt.show()
```

```python
# heirarchical 
# create dendrogram
linked=linkage(x_scaled,metric='euclidean',method="ward")
plt.figure(figsize=(20,10))
dendrogram(linked)
plt.show()
hc=AgglomerativeClustering(n_clusters=5)
hc_labels=hc.fit_predict(x_scaled)

print(silhouette_score(x_scaled, kmeans_labels))
```
