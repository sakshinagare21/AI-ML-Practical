```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.svm import SVC
from sklearn.metrics import classification_report,confusion_matrix
```

```python
df=pd.read_csv("UniversalBank.csv")
df.head(5)
```

```python
df.describe(include="all")
```

```python
df.dtypes
```

```python
df.info()
```

```python
# remove columns
print(df.columns)
# df=df.drop(columns=['ID'])
```

```python
df.isnull().sum()
```

```python
df=df.dropna(subset=['CreditCard'])
```

```python
X=df.drop(columns=['CreditCard','ZIP Code'])
```

```python
y=df['CreditCard']
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42,stratify=y)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
scaler=StandardScaler()
```

```python
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
model=SVC(kernel="linear",random_state=42,class_weight='balanced')

model.fit(x_train,y_train)
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import confusion_matrix,classification_report,precision_score,accuracy_score,r2_score
print("Accuracy:", accuracy_score(y_test, y_pred))

print("Precision:", precision_score(y_test, y_pred))

print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python
print("Actual values:")
print(y_test.value_counts())

print("Predicted values:")
print(pd.Series(y_pred).value_counts())
```

```python
print(y_train.value_counts())
print(y_test.value_counts())
```

```python
print(confusion_matrix(y_test, y_pred))
```
