```python
def find(parent,x):
    if parent[x]==x:
        return x
    else:
        parent[x]=find(parent,parent[x])
```

```python
def union(parent,rank,x,y):
    px=find(parent,x);
    py=find(parent,y);
    if px==py:
        return False
    if rank[px]>rank[py]:
        parent[py]=px
    elif rank[py]>rank[px]:
        parent[px]=py
    else:
        rank[px]+=1
        parent[py]=px
    return True
```

```python
def kruskal(graph,n):
    edges=[]

    for i in range(n):
        for j in range(n):
            if graph[i][j]!=0:
                edges.append((graph[i][j],i,j))

    parent=[x for x in range(n)]
    rank=[0]*n
    edges.sort()

    mst=[]
    total=0
    count=0
    for w,u,v in graph:
        if union(parent,rank,u,v):
            total+=w
            mst.append((w,u,v))
        if count==n-1:
            break
    if len(mst)==n-1:
        print(mst)
    
```
