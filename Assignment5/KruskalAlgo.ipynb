```python
def find(parent,x):
    if parent[x]==x:
        return x
    else:
        parent[x]=find(parent,parent[x])
```

```python
def union(parent,rank,x,y):
    px=find(parent,x);
    py=find(parent,y);
    if px==py:
        return False
    if rank[px]>rank[py]:
        parent[py]=px
    elif rank[py]>rank[px]:
        parent[px]=py
    else:
        rank[px]+=1
        parent[py]=px
    return True
```
