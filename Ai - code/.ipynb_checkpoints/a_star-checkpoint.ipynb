```python
import heapq

goal=[[1,2,3],
      [4,5,6],
      [7,8,0]]
def heuristic(graph):
    h=0

    for i in range(len(graph)):
        for j in range(len(graph[0])):

            if graph[i][j]!=0:

                x=(graph[i][j]-1)//3;
                y=(graph[i][j]-1)%3;

                # calculate the distance
                dis=abs(i-x)+abs(j-y)

                h+=dis
    return h
```

```python
def a_star(graph):
    pq=[]
    heapq.heappush(pq,(heuristic(graph),0,graph,[graph]))
    visited=set()

    while pq:

        f,g,curr,path=heapq.heappop(pq)

        if curr==goal:
            return path

        key=str(curr)

        if key in visited:
            continue
        visited.add(key)
        
        # find the position of blank 
        for i in range(len(curr)):
            for j in range(len(curr[0])):

                if curr[i][j]==0:

                    x,y=i,j

        # check for possible moves

        dir=[(0,1),(1,0),(-1,0),(0,-1)]

        for r,c in dir:

            nr=x+r;
            nc=y+c

            # check boundries

            if nr>=0 and nr<3 and nc>=0 and nc<3:

                # create new matrix
                new_matrix=[row[:] for row in curr]

                new_matrix[x][y],new_matrix[nr][nc]=new_matrix[nr][nc],new_matrix[x][y]

                if str(new_matrix) not in visited:

                    new_g=g+1

                    new_f=heuristic(new_matrix)+new_g

                    heapq.heappush(pq,(new_f,new_g,new_matrix,path+[new_matrix]))

    return None


graph = []

for i in range(3):
    row = list(map(int, input().split()))
    graph.append(row)

path = a_star(graph)

if path:
    print("Goal State Achieved")

    for mat in path:
        for row in mat:
            print(row)
        print()
else:
    print("No path found")

        
```

```python
# Implement the grid-based A* Search algorithm for obstacle avoidance and pathfinding. Display the optimal shortest path.

import heapq

def astar(graph,start,goal):
    row=len(graph)
    col=len(graph[0])

    pq=[(0,0,start,[start])]

    visited=set()

    while pq:
        f,g,curr,path=heapq.heappop(pq)

        if curr==goal:
            return path

        if curr in visited:
            continue
        visited.add(curr)

        r,c=curr

        # check to four direction

        dir=[(0,1),(1,0),(-1,0),(0,-1)]

        for dr,dc in dir:
            nr=dr+r
            nc=dc+c

            if 0<=nr<row and 0<=nc<col:
                new_pos=(nr,nc)
                if graph[nr][nc] == 0 and new_pos not in visited:
                    # heuristic distance 
                   dis=abs(goal[0]-nr)+abs(goal[1]-nc)

                   new_g=g+1

                   new_f=dis+new_g

                   heapq.heappush(pq,(new_f,new_g,new_pos,path+[new_pos]))
                
            
    return None


graph=[]

row=int(input("Enter the number of vertex"))

print("Enter the graph")

for i in range(row):
    r=list(map(int,input().split()))
    graph.append(r)

start=tuple(map(int,input("Enter start row and col").split()))
goal=tuple(map(int,input("Enter goal row and col").split()))


path=astar(graph,start,goal)

if path:
    print("Goal state reached")
    print(path)
else:
    print("Unable to reach goal state")
```

```python
# Implement the A* Search algorithm on a weighted graph using an admissible heuristic. Display the optimal path and total path cost.
import heapq
def astar(graph,heuristic,start,goal):
    pq=[(heuristic[start],0,start,[start])]

    visited=set()

    while pq:
        f,g,curr,path=heapq.heappop(pq)

        if curr==goal:
            return path,g

        if curr in visited:
            continue
        visited.add(curr)

        # check to four direction

        for neigh,weight in graph[curr]:
        
            if neigh not in visited:
                # heuristic distance 
                   new_g=g+weight

                   new_f=heuristic[neigh]+new_g

                   heapq.heappush(pq,(new_f,new_g,neigh,path+[neigh]))
                
            
    return None,-1

    
```

```python
graph={}

v=int(input("Enter the number of vertex:"))

for i in range(v):
    print("Vertex: ",i)
    val=input("Enter value of vertex:")
    graph[val]=[]

e=int(input("Enter the number of egdes: "))

for i in range(e):
    w,u,v=input("Enter weight,u,v: ").split()

    w=int(w)

    graph[u].append((v,w))
    graph[v].append((u,w))

print(graph)
heuristic={}
#enter heuristic distance
for vertex in graph:
    heuristic[vertex]=int(input(f"Enter heuristic value of {vertex}"))


start=input("Enter start vertex: ")
end=input("Enter goal vertex :")

path,cost=astar(graph,heuristic,start,end)

if path:
    print("Optimal Path:",path)
    print("Cost: ",cost)
else:
    print("No path")
```
