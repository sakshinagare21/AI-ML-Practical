```python
# Implement Prim's algorithm using a min-heap to construct the Minimum Spanning Tree (MST) of a weighted graph. Display the
# selected edges and total minimum cost.

import heapq

def prims(graph,start,n):

    pq=[(0,start,-1)]
    visited=[False]*n

    total=0
    count=0
    mst=[]

    while pq:
        cost,u,parent=heapq.heappop(pq)

        if visited[u]:
            continue
        total+=cost
        count+=1
        if parent!=-1:
            mst.append((cost,u,parent))
        visited[u]=True
        # check all neighbour
        for v in range(n):
            if graph[u][v]!=0 and not visited[v]:
                heapq.heappush(pq,(graph[u][v],v,u))
    # check is mst is possible are not
    if count!=n:
        print("Mst is not possible cycle exist")
    else:
        print("======Mst======")
        print(f"Total : {total}")
        for w,u,v in mst:
            print(f"{u} -> {v} : {w}")

graph=[]
v=int(input("Enter the number of vertex : "))
print("Enter the graph:")

for i in range(v):
    row=list(map(int,input().split()))
    graph.append(row)

print(graph)
start=int(input("Enter start Vertex: "))

prims(graph,start,v)

        
```
