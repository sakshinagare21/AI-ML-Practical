```python
def sort(a):

    for i in range(len(a)):
        min=i

        for j in range(i+1,len(a)):
            if a[j]<a[min]:
                min=j
        a[min],a[i]=a[i],a[min]
        print(f"Pass {i+1} :",a)
    print("Sorted Array: ",a)

a=[]
n=int(input("Enter the number of element :"))
print("Enter the array element:")
for i in range(n):
    a.append(int(input()))
sort(a)
```
