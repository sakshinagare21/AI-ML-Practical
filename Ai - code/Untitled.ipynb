```python
graph={}

v=int(input("Enter vertex:"))
for i in range(v):
    val=input(f"Enter value for vertex {i}")
    graph[val]=[]

e=int(input("Enter the number of edges:"))
for i in range(e):
    w,u,v=input("Enter w , u , v :").split()
    w=int(w)
    graph[u].append((v,w))
    graph[v].append((u,w))

print(graph)
```

```python
# dijkstra's algorithm
import heapq

def dijkstra(graph,start,v):
    pq=[(0,start)]
    parent={node:None for node in graph}
    dis={node:float('inf') for node in graph}

    dis[start]=0

    while pq:
        weight,u=heapq.heappop(pq)
        if dis[u]<weight:
            continue
        for neig,wei in graph[u]:
            new_dis=weight+wei
            if new_dis<dis[neig]:
                dis[neig]=new_dis
                parent[neig]=u
                heapq.heappush(pq,(new_dis,neig))
    for vertex in graph:
        if dis[vertex]==float('inf'):
            print(f"Unreachable {start} -> {vertex}")
        else:
            curr=vertex
            path=[]
            while curr!=None:
                path.append((curr))
                curr=parent[curr]
            path.reverse()
            print(path)
            print(f"Distance {start} -> {vertex} : {dis[vertex]}")


start=input("Enter start vertex")
dijkstra(graph,start,v)
```

```python
# prims

def prims(graph,start,n):
    visited={node:False for node in graph}
    pq=[(0,start,None)]
    mst=[]
    total=0
    count=0
    while pq:
        dis,u,parent=heapq.heappop(pq)

        if visited[u]:
            continue
        visited[u]=True
        total+=dis
        count+=1
        if parent!=None:
            mst.append((parent,u,dis))
        for neigh,wei in graph[u]:
            if not visited[neigh]:
                heapq.heappush(pq,(wei,neigh,u))
        
    if count==n:
        print("MST : ",mst)
        print(total)
    else:
        print("Cycle present")

start=input("Enter start vertex")
prims(graph,start,len(graph))  
```

```python
# krukals algo
vertex=[]
n=int(input("Enter vertex:"))
for i in range(n):
    val=input(f"Enter value for vertex {i}")
    vertex.append((val))

edges=[]
e=int(input("Enter the number of edges:"))
for i in range(e):
    w,u,v=input("Enter w , u , v :").split()
    w=int(w)
    edges.append((w,u,v))
print(vertex)
print(edges)
```

```python
def find(parent,x):
    if parent[x]==x:
        return x
    else:
         parent[x]=find(parent,parent[x])
         return parent[x]


def union(rank,parent,x,y):
    px=find(parent,x)
    py=find(parent,y)

    if px==py:
        return False
    if rank[px]>rank[py]:
        parent[py]=px
    elif rank[py]>rank[px]:
        parent[px]=py
    else:
        rank[px]+=1
        parent[py]=px
    return True


def krukals(edges,vertex):
    edges.sort()

    parent={node:node for node in vertex}
    rank={node:0 for node in vertex}

    total=0
    mst=[]
    count=0
    
    for w,u,v in edges:
        if union(rank,parent,u,v):
            count+=1
            total+=w
            mst.append((w,u,v))
        if count==len(vertex)-1:
            break
    if count==len(vertex)-1:
        print(mst)
        print(total)
    else:
        print("disconnected")

krukals(edges,vertex)
```

```python
# dijkstra's algorithm
import heapq

def dijkstra_grid(graph,start,n):
    pq=[(0,start)]
    parent=[-1]*n
    dis=[float('inf')]*n

    dis[start]=0

    while pq:
        weight,u=heapq.heappop(pq)
        if dis[u]<weight:
            continue
        for neig in range(n):
            if graph[u][neig]!=0:
                new_dis=weight+graph[u][neig]
                if new_dis<dis[neig]:
                    dis[neig]=new_dis
                    parent[neig]=u
                    heapq.heappush(pq,(new_dis,neig))
    for vertex in range(n):
        if dis[vertex]==float('inf'):
            print(f"Unreachable {start} -> {vertex}")
        else:
            curr=vertex
            path=[]
            while curr!=-1:
                path.append((curr))
                curr=parent[curr]
            path.reverse()
            print(path)
            print(f"Distance {start} -> {vertex} : {dis[vertex]}")


start=int(input("Enter start vertex"))
dijkstra_grid(grid,start,v)
```
