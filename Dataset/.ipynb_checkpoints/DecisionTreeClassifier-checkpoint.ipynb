```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("Heart.csv")
```

```python
print(df.head())
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
df.dtypes
```

```python
# manage null value
for col in df.columns:
    if df[col].dtypes=='object':
        df[col].fillna(df.mode().iloc[0])
    else:
        df[col].fillna(df.mean(numeric_only=True))
        
```

```python
from sklearn.preprocessing import LabelEncoder
# print(dir(preprocessing))
le=LabelEncoder()
```

```python
df['Thal']=le.fit_transform(df['Thal'])
```

```python
df['ChestPain']=le.fit_transform(df['ChestPain'])
```

```python

X=df.drop('AHD',axis=1)
y=df['AHD']
# remove the column
# Remove the Unnamed: 0 column
df = df.drop('Unnamed: 0', axis=1)
df=df.dropna(subset=['AHD'])
```

```python
from sklearn.tree import plot_tree
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier
```

```python
# perform encoding
df=pd.get_dummies(df,columns=['ChestPain','Thal'],drop_first=True)
```

```python
df['AHD']=df['AHD'].map({
    'No':0,
    'Yes':1
})
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
model=DecisionTreeClassifier(
    criterion="gini",max_depth=4,min_samples_split=2,random_state=42)
```

```python
# print(df.head(5))
print(x_train.select_dtypes(include='object').columns)
```

```python
model.fit(x_train, y_train)

y_pred = model.predict(x_test)
```

```python
print(df.select_dtypes(include='object').columns)
```

```python
from sklearn import metrics
print(dir(metrics))
```

```python
from sklearn.metrics import confusion_matrix,classification_report,precision_score,accuracy_score,r2_score
```

```python

print("Accuracy:", accuracy_score(y_test, y_pred))

print("Precision:", precision_score(y_test, y_pred))

print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
cm=confusion_matrix(y_test, y_pred)

sns.heatmap(cm,cmap="Blues",annot=True,fmt='g')
plt.title("Confusion Matrix")
plt.xlabel("Acutal Record")
plt.ylabel("Predicted Record")
plt.show()
```

```python
# show tree

plt.figure(figsize=(12,7))
plot_tree(model,max_depth=4,feature_names=X.columns,filled=True,rounded=True,fontsize=9)
plt.title("Decision Tree")
```
